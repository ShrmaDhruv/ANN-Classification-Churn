```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
model = load_model('model.h5')

# load scaler and encoder
with open('label_encoder_gender.pkl','rb') as file:
    label_encoder_gender = pickle.load(file)
with open('Scaler.pkl','rb') as sc:
    scaler = pickle.load(sc)
with open('OneHotEncoder_Geography.pkl','rb') as f:
    Geo = pickle.load(f)
```

```python
input_data_3 = {
    "CreditScore": 810,
    "Geography": "Spain",
    "Gender": "Male",
    "Age": 29,
    "Tenure": 8,
    "Balance": 0.0,
    "NumOfProducts": 2,
    "HasCrCard": 1,
    "IsActiveMember": 1,
    "EstimatedSalary": 98000.00
}
```

```python
input_data_2 = {
    "CreditScore": 420,
    "Geography": "France",
    "Gender": "Female",
    "Age": 54,
    "Tenure": 1,
    "Balance": 145000.75,
    "NumOfProducts": 1,
    "HasCrCard": 0,
    "IsActiveMember": 0,
    "EstimatedSalary": 62000.50
}
```

```python
geo_encoded = Geo.transform([[input_data_2['Geography']]])
```

```python
geo_encoded
```

```python
geo_encoded_df = pd.DataFrame(geo_encoded,columns=Geo.get_feature_names_out())
```

```python
geo_encoded_df
```

```python
df1 = pd.DataFrame([input_data_2])
```

```python
df1
df1 = df1.drop('Geography',axis=1)
```

```python
df1 = pd.concat([df1,geo_encoded_df],axis=1)
```

```python
df1
```

```python
df1['Gender'] = label_encoder_gender.transform(df1['Gender'])
df1
```

```python
input_scaled = scaler.transform(df1)
```

```python
input_scaled
```

```python
prediction = model.predict(input_scaled)
```

```python
prediction_proba = prediction[0][0]
```
